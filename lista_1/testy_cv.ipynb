```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
monety_szare = cv2.imread('../data/Monety_szare.bmp')
monety_szum = cv2.imread('../data/Monety_szare_szum.bmp')
```

```python
import cv2
import matplotlib.pyplot as plt

th_type = [cv2.THRESH_BINARY, cv2.THRESH_BINARY_INV, cv2.THRESH_TRUNC, cv2.THRESH_TOZERO]
Ts = 125

fig, axes = plt.subplots(2, 2, figsize=(10, 12))
axes = axes.flatten()

for ax, t in zip(axes, th_type):
    T, img_threshInv = cv2.threshold(monety_szare, Ts, 255, t)
    ax.imshow(img_threshInv, cmap='gray', vmin=0, vmax=255)
    ax.set_title(f'Rodzaj progowania: {t:.0f}')
    ax.axis('off')

plt.tight_layout()
plt.show()
```
